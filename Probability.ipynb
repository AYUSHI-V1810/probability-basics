**Question 2: In a class of 50 students:**    

20 like Mathematics (M)

15 like Science (S)

5 like both subjects

What is the probability that a student chosen at random likes Mathematics or Science?  

ANSWER  
- Total students = 50
- Maths = 20
- Science = 15
- Both = 5

Students liking Maths or Science = Maths + Science - Both = 20 + 15 - 5 = 30

Probability = 30/50 = 3/5 = 0.6 or 60%

So, the probability is 3/5 or 60%

**Question 3: A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability

**Question 4: The population of a school is divided into 60% boys and 40% girls. If you want equal representation
of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling?
Why?**

ANSWER -   
WE USE **STRATIFIED SAMPLING.**    
- Ensures representation of both boys (60%) and girls (40%) in the sample
- Maintains the population's gender proportion
- Reduces sampling bias

Simple Random Sampling might lead to uneven representation, but Stratified Sampling guarantees equal or proportionate representation of both genders.

**Question 5: The average height of 1000 students = 160 cm. A sample of 100 students shows an average height
= 158 cm.**

Find the sampling error.

ANSWER -   
- Population mean (a) = 160 cm
- Sample mean (x) = 158 cm

Sampling Error = x - a = 158 - 160 = -2 cm

So, the sampling error is -2 cm.

**Question 6: The population mean salary is ₹50,000 with σ = ₹5,000. If we take a sample of 100 employees, what
is the standard error of the mean (SEM)?**

ANSWER  -  
- Population std dev (σ) = 5000
- Sample size (n) = 100

Standard Error of the Mean (SEM) = σ / √n
= 5000 / √100
= 5000 / 10
= 500

**Question 8: From a deck of 52 cards, two cards are drawn without replacement. What is the probability that
both are Aces?**

ANSWER -  
In a standard deck of 52 cards there are 4 Aces.

Drawing 2 cards without replacement:

- P(1st card is Ace) = 4/52 = 1/13
- P(2nd card is Ace) = 3/51 = 1/17

Joint probability:   
P(BOTH ACES) =  (1/13) * (1/17)  = 1/221  
Final answer → 1/221 (≈ 0.00452 or 0.45%)

**Question 10: Differentiate between discrete and continuous random variables with examples.**

ANSWER  -  
- Discrete Random Variable: Takes countable, distinct values (often integers)
    - Example: Number of heads in 3 coin tosses (0, 1, 2, 3)
    - Example: Number of students in a class (can't be 2.5)
- Continuous Random Variable: Takes any value in a range (including decimals)
    - Example: Height of students (can be 170.5 cm, 171.23 cm...)
    - Example: Time taken to run 100m (10.2s, 10.25s...)

**Key difference: Discrete → countable; Continuous → measurable.**
